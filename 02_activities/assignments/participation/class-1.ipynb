### Import the necessary libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
np.random.seed(613)
x = np.arange(50)
# provide fifty random integers between 0 and 100
y = np.random.randint(0, 100, 50) 
```

```python
x # x-axis values
```

```python
y # y-axis values
```

### Visualization

```python
# figsize=(5,3) sets the width to 5 inches and height to 3 inches
fig, ax = plt.subplots(figsize=(5,3)) 
# scatter plot
ax.scatter(x, y) 
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x,y) # bar plot
```

### Add Title and labels

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y) # Plot a line chart of y values versus x values

ax.set_title("Total growth over times") 
# Label the x-axis to describe what the horizontal axis represents
ax.set_xlabel("Total growth")
# Label the y-axis to describe what the vertical axis represents
ax.set_ylabel("years since start")   
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y, bins=10) # histogram with 10 bins
```

### Define custom font settings for text elements in the plot

```python
# font1 is a dictionary of font properties for titles
font1 = {'family': 'sans-serif', 'color': 'blue', 'size': 20 }
# font2 is a dictionary of font properties for axis labels or other text
font2 = {'family': 'monospace', 'color': 'green', 'size': 11 }
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

# Set the title for the plot with custom font and alignment
ax.set_title("Total growth over time", 
             fontdict=font1, # Apply the custom font style defined earlier
             loc= 'left'# Position the title on the left side of the chart)
ax.set_ylabel("Total growth", fontdict=font2)
ax.set_xlabel("years since start", fontdict=font2)  
```

```python
# libraries (included here so this cell can run standalone)

import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
import numpy as np
import pandas as pd


# dataframe
df = pd.DataFrame({
'var1': np.random.normal(size=1000),
'var2': np.random.normal(loc=2, size=1000) * -1
})

# Fig size
plt.rcParams["figure.figsize"]=12,8

# plot density chart for var1
sns.kdeplot(data=df, x="var1",  fill=True, alpha=1)

# plot density chart for var2
kde = gaussian_kde(df.var2)
x_range = linspace(min(df.var2), max(df.var2), len(df.var2))

# multiply by -1 to reverse axis (mirror plot)
sns.lineplot(x=x_range*-1, y=kde(x_range) * -1, color='orange') 
plt.fill_between(x_range*-1, kde(x_range) * -1, color='orange')

# add axis names        
plt.xlabel("value of x")
plt.axhline(y=0, linestyle='-',linewidth=1, color='black')

# show the graph
plt.show()
```
