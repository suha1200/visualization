```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import scipy.stats as stats
import PIL
import requests
```

```python
np.random.seed(613)
```

```python
tips = sns.load_dataset("tips")
print(tips.head())
```

```python
sns.set_style('whitegrid')
tipgraph = sns.lineplot(data=tips,       # choose our dataset
           x='total_bill',  # define our x variable
           y='tip')         # define our y variable
tipgraph.set(title = 'Tips vs Total Bill',
xlabel = 'Total Bill ($)',
ylabel = 'Tip Amount ($)')
plt.show()
```

```python
fig = plt.subplots(figsize=(10,3))

tipgraph = sns.lineplot(data=tips,
                      x='total_bill',
                      y='tip',
                      color = 'hotpink',
                      linestyle = '--',
                      linewidth = 3,
                      marker = 'o',
                      markerfacecolor = 'indigo')
tipgraph.set(title='Tips vs. Total Bill',
            xlabel='Total Bill ($)',
            ylabel='Tip Amount ($)')
plt.show()
```

```python
tipgraph = sns.scatterplot(data=tips,
                         x='total_bill',
                         y='tip',
                         style = 'time',
                         hue = 'day',
                         palette =['green', 'orange', 'purple','black','yellow', 'pink','brown'],)
tipgraph.set(title='Tips vs. Total Bill',
            xlabel='Total Bill ($)',
            ylabel='Tip Amount ($)')
plt.show()
```

```python
sns.pairplot(
      data = tips,
      hue = 'day')
plt.show()
```

```python
# Create a relational plot (relplot) using seaborn to visualize the relationship between total_bill and tip
daysplot = sns.relplot(
    data=tips,          # Use the 'tips' dataset
    x="total_bill",     # Set 'total_bill' as the x-axis variable
    y="tip",            # Set 'tip' as the y-axis variable
    hue="sex",          # Use different colors to represent male and female
    col="day",          # Create separate plots (columns) for each day of the week
    kind="scatter",     # Use a scatter plot to show individual data points
    col_wrap=2          # Wrap plots into multiple rows with 2 plots per row
)
```

```python
import plotly.graph_objects as go     # 'go' is 'graph objects’
import nbformat
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))

graph.update_layout(title='One Piece Crew Power Levels',
                   xaxis_title='Crew Members',
                   yaxis_title='Power Level',
                   width=700,   # set width (in pixels)
                   height=400   # set height (in pixels))
)
graph.show()
```

```python
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers',    # we want points for a scatter plot
    marker=dict(
        size=15,                            # point size
        color='hotpink',                    # point colour
        opacity=1,                          # point transparency/alpha
        line=dict(width=5, color='purple')  # point outline
    )))
graph.update_layout(
    title='Interactive Pirate Plot',
    xaxis_title='Pirates',
    yaxis_title='Scores',
    width=500, height=500)
```

```python
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
                  on_bad_lines='skip')
df
```

```python
#!pip install wordcloud
```

```python
# join all our text from each row from our quote column into a string
text = " ".join(each for each in df.quote)
# generate our wordcloud image
wordcloud = WordCloud(background_color="white",
                      colormap = 'inferno',
                      random_state = 613).generate(text)
# use matplotlib syntax to put our image in a figure
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud,                  # remember 'imshow' from when we added pictures to our matplotlib axes
          interpolation='bilinear'
          )   # this line helps smooth our image
ax.axis("off")
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
```

```python
#!pip install matplotlib-venn
```

```python
A = set(["apple", "banana", "watermelon", "blueberry", "peach"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
diagram = venn2_unweighted([A, B],
                            set_labels = ('Fruits', 'Pies'),
                            set_colors=("blue", "red"),
                            alpha=0.5)

diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))

plt.show()
```
