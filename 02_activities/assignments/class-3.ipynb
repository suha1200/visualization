```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import requests
import PIL
from io import BytesIO
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
# ax.plot(x,y1,y2) Or
ax.plot(x,y1, label = "Person1")
ax.plot(x,y2, label = "Person2")
ax.legend(loc="lower right",
          frameon=True,
          fontsize = 12,
          ncol=2,
          shadow=True)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
# ax.plot(x,y1,y2) Or
ax.plot(x,y1, label = "Person1")
ax.plot(x,y2, label = "Person2")
ax.text(10, 95, "This value is important!")

ax.legend(loc="upper left", # lower left, upper right, best
          frameon=True, 
          bbox_to_anchor=(1,1)
)
```

### Session-2

```python
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
plt.style.use('ggplot')
fig, (ax1,ax2) = plt.subplots(ncols=2,
                              nrows=1,
                              figsize=(10,4))
ax1.scatter(x1,y1, label = "First Plot!")
plt.xlabel('First Plot!', fontsize = 18, fontdict = font1)

ax2.bar(x2, y2, label="Second Plot!")
plt.xlabel('Second Plot!', fontsize = 18, fontdict = font1)

```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                     ['ax2','ax3']],
                                    figsize=(8,6),
                                    layout = "constrained")
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
plt.suptitle("My Subplot Mosaic!", fontsize=22)

```

```python
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(x,y1,color='indigo')
ax.plot(x,y2,color='red')
```

```python
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x,y2,color='brown')
ax.errorbar(x, y2, 
            yerr=y2_sd, fmt='o',
            ecolor= "orange",
            elinewidth= 4,
            capsize = 6,
            capthick= 4)
```
